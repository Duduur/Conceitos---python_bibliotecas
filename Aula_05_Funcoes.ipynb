## 1) O que é uma função

Uma função é um bloco de código nomeado que representa uma transformação ou ação reutilizável.

Pense em função como um contrato:

- **Entrada**: parâmetros, que são os dados que entram na função.
- **Processamento**: o corpo da função, que executa uma lógica.
- **Saída**: o valor retornado (ou a ausência de retorno, quando a função só produz efeito como imprimir).

Por que funções existem:

- **Reuso**: evita duplicação de lógica.
- **Organização**: divide um problema grande em partes menores.
- **Testabilidade**: cada parte pode ser testada isoladamente.
- **Legibilidade**: nomes bons substituem comentários longos.
- **Manutenção**: uma correção em um lugar melhora todo o sistema.



## 2) Anatomia de uma função em Python

Estrutura mínima:

```python
def nome_da_funcao(parametros):
    """Docstring: descreve o contrato da função."""
    corpo
    return saida
```

Detalhes importantes:

- `def` cria a função.
- `nome_da_funcao` é o identificador usado para chamar a função.
- `parametros` são variáveis locais que recebem valores na chamada.
- A docstring é um texto opcional usado por ferramentas e pelo `help()`.
- `return` devolve um valor ao chamador e encerra a execução da função naquele ponto.
- Se não houver `return`, o retorno é `None`.

Função não executa quando é definida, ela só executa quando é chamada.


````python
def area_retangulo(base, altura):
    """ Calculo da area de um triãngulo
    Parâmetros:
        base(float): medida da base
        altura(dloat): medida da altura
    Retorna:
        float: área calculada
    """
    area = base * altura # processamento
    return area # saida
````

````python
print(area_retangulo(3,5))
````


## 3) Parâmetros: posicionais e nomeados

Quando você chama uma função, pode passar argumentos de dois jeitos:

- **Posicional**: pela ordem.
- **Nomeado**: pelo nome do parâmetro.

Argumentos nomeados aumentam legibilidade e reduzem erro de ordem.


````python
def saudacao(nome, curso):
    "Monta uma saudação simples"
    return f"Ola, {nome}. Bem-vinda(o) ao curso de {curso}."
````

````python
print(saudacao("Bruna", "Inteligencia de Dados"))
````


## 4) Valores padrão: parâmetros opcionais

Um parâmetro com valor padrão se torna opcional na chamada.

Regras práticas:

- Parâmetros obrigatórios primeiro, opcionais depois.
- O padrão deve ser seguro e coerente com a função.


````python
def formatar_preco(valor, moeda= "R$"):
    "Formata um valor monetario com simbolo"
    return f"{moeda} {valor:.2f}".replace(",","x").replace(".",",").replace("x",".")
````

````python
print(formatar_preco(1500))
````


## 5) Retorno, None e efeitos colaterais

Há dois estilos comuns de função:

- **Função com retorno**: devolve um valor, ideal para cálculo e testes.
- **Função com efeito colateral**: imprime, grava arquivo, altera estado, ideal para interação e I/O.

Ponto didático importante:

- Prefira retornar valores e deixar `print()` fora das funções, exceto em funções de interface.


````python
def celsius_para_fahrenheit(c):
    "Converte a temperatura de Celsius para Fahrenheit"
    return(c*9/5)*32 #Retorna a conversão
````

````python
temp_f = celsius_para_fahrenheit(25)
print("Fahrenheit:", temp_f) # saida fora da função
````


## 6) Validação de entrada e uso de exceções

Funções profissionais costumam validar entradas. Duas abordagens:

- Retornar um valor especial, como `None`.
- Lançar uma exceção, o que interrompe o fluxo e obriga o chamador a tratar.

Nesta aula, use exceções simples para sinalizar contrato violado.


````python
def dividir(a,b):
    """ Divide a poor b 
    Levanta:
        ValueError: se b for 0
    """
    if  b == 0: 
        raise ValueError("Divisão por zero não é permitida") # Protege o contrato
    return a/b # Retorna a divisão
````

````python
print(dividir(10, 2))
````

````python
try:
    print(dividir(10, 0)) # Gera o erro
except ValueError as e:
    print("Erro: ", e) # Trata erro
````


## 7) Escopo: variável local, global e tempo de vida

Dentro de uma função, variáveis criadas são **locais** por padrão.

- Variável local existe apenas durante a execução da função.
- Variável global existe no módulo inteiro.

Boas práticas:

- Evite depender de globais.
- Prefira passar tudo por parâmetro e retornar resultados.


````python
taxa = 0.1 # Variavel global

def aplicar_taxa(valor):
    "Aplica a taxa global a u valor"
    return valor *(1+taxa) # LÊ GLOBAL, NÃO ALTERA
````

````python
print(aplicar_taxa(100))
````

````python
def aplicar_taxa_com_paramentro(valor, taxa_local):
    "Aplica a taxa recebida por parâmetro"
    return valor * (1+taxa_local) # mais previsivel
````

````python
print(aplicar_taxa_com_paramentro(100,taxa_local=0.02))
````


## 8) Mutabilidade e armadilha de padrão mutável

Este é um ponto clássico de Python.

Não use lista ou dicionário como valor padrão de parâmetro, porque o objeto é criado uma vez e reaproveitado.

Use `None` e inicialize dentro da função.


````python
def adicionar_item(item, lista= None):
    "Adicionar item a uma lista, evitando padrão mutável"
    if lista is None:
        lista = [] #Cria uma nova lista por chamada
    lista.append(item)#altera lista local
    return lista
````

````python
print(adicionar_item("B"))
````

````python
base = ["A"]
print(adicionar_item("B", base))
````


## 9) Múltiplos retornos e desempacotamento

Uma função pode retornar mais de uma informação ao mesmo tempo. Em Python isso é feito retornando uma tupla.

Isso é útil para retornar resultado e metadados sem criar estruturas complexas.


````python
def resumo_numeros(a,b):
    "Retorna soma, diferença e produto de dois números"
    soma = a+b
    diferenca = a-b 
    produto = a*b
    return soma, diferenca, produto
````

````python
s, d, p =resumo_numeros(10,3) #desempcota
print("Soma:", s)
print("Diferença:", d)
print("Produto:", p)
````

````python
j = resumo_numeros(10,3)
````


## 10) Funções como unidades de teste: asserts simples

Sem usar `for` ou `while`, ainda é possível criar testes diretos com `assert`.

Se um `assert` falhar, o Python levanta `AssertionError` e aponta o problema.


````python
def eh_maior_de_idade(idade):
    "Retorna True se idade indicar maioridade"
    return idade>=18
````

````python
assert eh_maior_de_idade(18) is True # Teste de fronteira
assert eh_maior_de_idade(17) is False # Teste abaixo
print("Testes básicos passaram")
````


## 11) Usando bibliotecas: quando reinventar não vale a pena

Python tem bibliotecas padrão com funções prontas e bem testadas.

Exemplos úteis:

- `math` para cálculos
- `statistics` para média, mediana, desvio padrão

Aqui, vamos usar `statistics.mean` para evitar escrever lógica repetida.


````python
import statistics #Biblioteca padrão

def media_3_notas(n1,n2,n3):
    "Calcula a media de três notas usando biblioteca"
    return  statistics.mean([n1,n2,n3]) #Calcula a media
````

````python
m = media_3_notas(7,8,9)
print("Média:", m)
````


## 12) Mini projeto sem laços: verificador simples de aprovação

Regras:

- Entrada: três notas
- Processamento: média
- Saída: aprovado se média >= 6

Objetivo didático:

- decompor em funções pequenas
- separar cálculo, decisão e apresentação


````python
import statistics

def ler_3_notas():
    "Lê três notas do usuário retorna como tupla"
    n1 = float(input("Nota 1: ")) # entrada 1
    n2 = float(input("Nota 2: ")) # entrada 2
    n3 = float(input("Nota 3: ")) # entrada 3
    return n1, n2, n3 #Retorna entradas
    
def calcular_media(n1, n2, n3):
    "Calcular a média de 3 valores"
    return statistics.mean([n1,n2,n3]) # Processamento

def est_aprovado(media, corte = 6.0):
    "Decide aprovação  pela média e nota de corte"
    return media >= corte # Decisão

def apresentar_resultado(media, aprovado):
    "Mostra o resultado final ao usuário"
    status ="Aprovado" if aprovado else "Reprovado" # Mostra texto conforme decisão
    print("Média", round(media,2)) # Saida
    print("Status", status) #Saida
````

````python
n1, n2, n3 = ler_3_notas()# entrada
media = calcular_media(n1, n2 ,n3) # processamento
aprov = est_aprovado(media)#dicisão
apresentar_resultado(media,aprov)
````

## 2) Função de desconto

Crie `aplicar_desconto(valor, percentual=10)` que retorna o valor com desconto.

````python
valor = float(input("Digite o valor da compra a pagar"))

def aplicar_desconto(valor, percentual =10):
    if valor > 50:
        desconto = percentual / 100
        desconto = valor * desconto
        valor = valor - desconto
        return  valor
    else:
        return "Sem desconto disponivel"
    
print(aplicar_desconto(valor, percentual=10))
````

Crie `validar_score(score)` que levanta `ValueError` se score estiver fora de 0 a 1000.

````python
score = int(input("Digite seu score (0 a 1000)"))

def validar_score(score):
    
    if score >= 0 and score <=1000:
        return score, "Score validado com sucesso"
    else:
        raise ValueError("Valor do Score está fora do intervalo de 0  1000") # Protege o contrato   
   
````
