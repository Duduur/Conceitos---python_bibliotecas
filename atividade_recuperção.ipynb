## 3. Dicionário — Definição e Estrutura Interna

**Definição:** Um dicionário armazena pares **chave:valor**

- Cada entrada é um par `chave: valor` separado por vírgula.
- Chaves devem ser únicas e imutáveis (strings, números, tuplas).
- Valores podem ser de qualquer tipo.

```python
tupla = {"nome": "Eduuu", "idade": 20,"saldo":2.50, "Situação": True}
print(tupla)
print(tupla["nome"])
```

## 4. Mutabilidade 

**Definição:** Mutabilidade é  quamdo um objeto ser pode ser **alterado após sua criação na memória**.

- **Lista** → mutável: `append()`, `remove()`, mudança nos índices sáo permitidas.
- **Tupla** → imutável: qualquer tentativa de modificação gera um erro.


## 5. Variável Local vs. Global em Funções

- **Local:** declarada dentro da função; existe apenas durante a execução da função.
- **Global:** declarada fora de funções; acessível por todo o codigo. 

```python
varialvel_global = 10

def func():
    variavel_local = 5
    print(variavel_local)

func()
print(varialvel_global)
```

## 6. Escopo de Variáveis

**Definição:** Escopo é a região do código onde uma variável pode ser referenciada
Variaveis dentro da função só existem nela


## 7. Função

**Definição:**  É um bloco de código que é reutilizando durante o codigo, que executa uma tarefa específica.


```python

def calcular_area(base, altura):    # declaração
    area = base * altura            # corpo
    return area                     # retorno

resultado = calcular_area(5, 3)     # chamada com argumentos
print('Área:', resultado)           # 15
```

## 8. Parâmetro vs. Argumento

- **Parâmetro:** variável declarada na *assinatura* da função (na definição).
- **Argumento:** valor concreto passado na *chamada* da função.

```python
def saudar(nome):         # 'nome' é o PARÂMETRO
    print(f'Olá, {nome}!')

saudar('Maria')            # 'Maria' é o ARGUMENTO
saudar('João')             # 'João' é o ARGUMENTO
```

## 9. Retorno de uma Função

**Definição:** O `return` devolve um resultado. 
Sem ele, a função retorna `None`. É possível retornar múltiplos valores (como tupla).

```python
def somar(a, b):
    return a + b

def sem_retorno():
    print('executa')

x = somar(3, 7)         # x recebe 10
y = sem_retorno()       # y recebe None
print(x, y)             # 10 None
```

## 13. Classificar uma Nota (A, B, C, D)

```python
nota = float(input('Digite a nota (0 a 10): '))

if nota >= 9:
    conceito = 'A'
elif nota >= 7:
    conceito = 'B'
elif nota >= 5:
    conceito = 'C'
else:
    conceito = 'D'

print(f'Nota: {nota:.1f} — Conceito: {conceito}')
```

## 15. Verificar se um Número é Múltiplo de 3 e 5 Simultaneamente

```python
numero = int(input('Digite um número: '))

if numero % 3 == 0 and numero % 5 == 0:
    print(f'{numero} é múltiplo de 3 E de 5 (múltiplo de 15)')
elif numero % 3 == 0:
    print(f'{numero} é apenas múltiplo de 3')
elif numero % 5 == 0:
    print(f'{numero} é apenas múltiplo de 5')
else:
    print(f'{numero} não é múltiplo de 3 nem de 5')
```

## 16. Diferença entre for e while

For = Usado quando você já sabe qauntas vezes vai repetir, geralmente usado para sequencia

While = Usado quando a repetição depende de uma condição, pode repetir indefinidamente se a condição nunca for falsa


```python
# for 
for i in range(5):
    print('for:', i)

# while 
i = 0
while i < 5:
    print('while:', i)
    i += 1
```

## 17. Imprimir Números de 1 a 10 com `for`

## 18. Imprimir Números de 1 a 10 com `while`

```python
numero = 1

while numero <= 10:
    print(numero)
    numero += 1    # incrementa o contador
```

## 23. Remover Valores Duplicados de uma Lista

```python
lista = [1, 2, 3, 2, 4, 1, 5, 3, 6]
lista_2 = []

for item in lista:
    for x in lista_2:
        if item == x:
            break
    else:
        lista_2.append(item)

print('Original:', lista)
print('Sem duplicatas:', lista_2)
```

## 30. Separar uma Frase em Palavras com Estrutura de Repetição

```python
frase = input('Digite uma frase: ')

palavra = ''

for letra in frase:
    if letra != ' ':
        palavra += letra
    else:
        print(palavra)
        palavra = ''


if palavra != '':
    print(palavra)
```

## 31. Função que Retorna o Maior entre Dois Números

```python
def maior(a, b):
    if a > b:
        return a
    elif b > a:
        return b
    else:
        return 'Os números são iguais'

print(maior(10, 25))   
print(maior(7, 3))     
print(maior(5, 5))     
```

## 32. Função que Calcula o Fatorial de um Número

```python
def fatorial(n):
    resultado = 1
    
    for i in range(1, n + 1):
        resultado = resultado * i
    
    return resultado

# teste
num = int(input('Digite um número: '))
print('Fatorial:', fatorial(num))
```

## 33. Função que Recebe uma Lista e Retorna sua Soma

```python
def soma_lista(lista):
    soma = 0
    
    for n in lista:
        soma += n
    
    return soma


valores = [1, 2, 3, 4]
print('Soma:', soma_lista(valores))
```

## 34. Função que Verifica se um Número é Primo

```python
def eh_primo(n):
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
    
    return True


num = int(input('Digite um número: '))

if eh_primo(num):
    print('É primo')
else:
    print('Não é primo')
```
