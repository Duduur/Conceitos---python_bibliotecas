### Passo 1: Carregamento direto com pandas:
  

```python
import pandas as pd

url ='https://catalog.ourworldindata.org/garden/covid/latest/compact/compact.csv'
df = pd.read_csv(url)

```

#### 6) Filtre os dados apenas para o Brasil

```python
brasil =df[df['country'] == 'Brazil']
brasil.tail(2)
```

#### 7) Exiba o intervalo de datas disponíveis para o Brasil

```python
brasil['date'].min(),brasil['date'].max()
```

#### 8) Qual o pico de novos casos diários no Brasil e quando ocorreu?

```python
idx_pico = brasil['new_cases'].idxmax()
brasil.loc[idx_pico, ['date', 'new_cases']]
```

#### 9) Crie um gráfico com os novos casos registrados diariamente no Brasil

```python
import matplotlib.pyplot as plt

brasil.set_index('date')['new_cases'].plot(figsize=(12,6), title= 'Novos Casos Diários no Brasil')
plt.xlabel('Data')
plt.ylabel('Novos Casos')
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### 10) Crie um gráfico com os novos casos acumulados registrados diariamente no Brasil

```python
brasil.set_index('date')['total_cases'].plot(figsize=(10,6), title='Casos de Covid-19 no Brasil')
plt.ylabel("Numero de casos")
plt.xlabel("Data")
plt.grid =True
plt.tighl_layout() #Ajuda automaticamente o Layout para evitar sobreposição
plt.show()
```

#### 11) Quantos dias o Brasil teve mais de 1000 mortes diárias?

```python
print("No Brasil ocorreu mais de 1000 mortres em ",
      len(brasil[brasil['new_cases']>1000]), "dias")
```

#### 12) Calcule a taxa de letalidade por país

```python
df_letalidade = df.groupby('country')[['total_deaths','total_cases']].max()
df_letalidade['letalidade'] = ((df_letalidade['total_deaths']/df_letalidade['total_cases'])*100).round(2)
df_letalidade.sort_values(by='letalidade',ascending=False).head(30)
```

### Desafio Final
##### Crie um Dashboard simples, em que se pode selecioanar um determinado país e ver as principais informações sobre a Covid-19.

```python
plt.figure(figsize=(8,6))
plt.bar(brasil['new_cases'],
         bins=30,
         edgecolor='black',
         color='orange',
         alpha=0.7)

plt.title('Distribuição de Novos Casos')
plt.xlabel('Casos')
plt.ylabel('Frequência')
plt.show()
```
